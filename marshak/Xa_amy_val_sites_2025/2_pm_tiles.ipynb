# Organizing

```python
tif_data = [{'dist_type': dist_token.split('__')[0],
             'mgrs_tile_id': dist_token.split('__')[1],
             'dist_s1_gen_status': dist_s1_prod_path,
             'dist_hls_veg_status': dist_veg_hls_prod_path,
             'dist_hls_gen_status': dist_gen_hls_prod_path}
             for (dist_token, dist_s1_prod_path, dist_gen_hls_prod_path, dist_veg_hls_prod_path) in zip(dist_types, 
                                                                                 dist_s1_gen_status_paths,
                                                                                 dist_hls_gen_status_paths,
                                                                                 dist_s1_veg_status_paths)]
tif_data[:3]
            
```

```python
import shutil

shareable_data = Path('gen_status_examples')
shareable_data.mkdir(exist_ok=True, parents=True)
```

```python
for item, token in zip(tif_data, dist_types):
    out_dir = shareable_data / token
    out_dir.mkdir(exist_ok=True, parents=True)
    
    src_dist_s1 = item['dist_s1_gen_status']
    shutil.copy(str(src_dist_s1), str(out_dir / src_dist_s1.name))

    src_dist_hls_gen = item['dist_hls_gen_status']
    shutil.copy(src_dist_hls_gen, out_dir / src_dist_hls_gen.name)

    src_dist_hls_veg = item['dist_hls_veg_status']
    shutil.copy(src_dist_hls_veg, out_dir / src_dist_hls_veg.name)
```
