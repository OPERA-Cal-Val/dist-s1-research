```python
from pathlib import Path
import shutil
```

```python
out_dir = Path("out_hard")
assert out_dir.exists()
```

```python
all_pdf_dir = Path(f'all_pdfs_{out_dir.name}')
all_pdf_dir.mkdir(parents=True, exist_ok=True)
```

```python

pdf_paths = list(out_dir.rglob("*.pdf"))
pdf_paths[:3]
```

```python
[shutil.copyfile(p, all_pdf_dir / p.name) for p in pdf_paths]
```
